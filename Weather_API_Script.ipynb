```python
import requests
import pandas as pd

cities = [
    {"city" : "Berlin","country" : "Germany","lat":52.52,"lon" : 13.41},
    {"city":"Paris","country":"France","lat":48.85,"lon":2.35},
    {"city":"Rome","country":"Italy","lat":41.90,"lon":12.49},
    {"city":"Madrid","country":"Spain","lat":40.41,"lon":-3.70},
    {"city":"Amsterdam","country":"Netherlands","lat":52.37,"lon":4.89},
    {"city":"Vienna","country":"Austria","lat":48.20,"lon":16.37}
]

all_data = []
for city_info in cities:
  url = (
      f"https://api.open-meteo.com/v1/forecast?"
        f"latitude={city_info['lat']}"
        f"&longitude={city_info['lon']}"
        f"&hourly=temperature_2m"
  )
  response = requests.get(url)
  data = response.json()
  temp_df = pd.DataFrame({
      "city": city_info["city"],
        "country": city_info["country"],
        "time": data["hourly"]["time"],
        "temperature_2m": data["hourly"]["temperature_2m"]
  })

  all_data.append(temp_df)
  weather_df = pd.concat(all_data, ignore_index=True)
  weather_df.head()
  print(weather_df)
```

```python
weather_df.to_csv(
    "weather_data.csv",index=False
)
print("CSV CREATED")
```
